# **Random Forest Data Science Project**

```python
import pandas as pd
import numpy as np
import joblib
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import OrdinalEncoder
import os
from google.colab import files
```

```python
FILES = {
    1: 'bronx_2021.xlsx',
    2: 'brooklyn_2021.xlsx',
    3: 'manhat_2021.xlsx',
    4: 'queens_2021.xlsx',
    5: 'staten_2021.xlsx'
}

BOROUGH_NAMES = {1: 'Manhattan', 2: 'Bronx', 3: 'Brooklyn', 4: 'Queens', 5: 'Staten Island'}

VALID_CATEGORIES = [
    '01 ONE FAMILY DWELLINGS',
    '02 TWO FAMILY DWELLINGS',
    '03 THREE FAMILY DWELLINGS',
    '07 RENTALS - WALKUP APARTMENTS',
    '11 SPECIAL CONDO BILLING LOTS'
]

FEATURES = ['GROSS SQUARE FEET', 'LAND SQUARE FEET', 'YEAR BUILT', 'ZIP CODE', 'BUILDING CLASS CATEGORY']
```

```python
if not os.path.exists('models_output'):
  os.makedirs('models_output')

for borough_code, file_path in FILES.items():
  print(f'Working on {BOROUGH_NAMES[borough_code]}')
  try:
    df = pd.read_excel(file_path, skiprows=4)
    df_clean = df[
            (df['SALE PRICE'] > 10000) &
            (df['RESIDENTIAL UNITS'] > 0) &
            (df['COMMERCIAL UNITS'] == 0) &
            (df['GROSS SQUARE FEET'] > 0) &
            (df['YEAR BUILT'] > 0) &
            (df['BUILDING CLASS CATEGORY'].isin(VALID_CATEGORIES))
        ].copy()
    x = df_clean[FEATURES]
    y = np.log1p(df_clean['SALE PRICE'])

    encoder = OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1)
    x[['BUILDING CLASS CATEGORY']] = x[['BUILDING CLASS CATEGORY']].astype(str)
    x[['BUILDING CLASS CATEGORY']] = encoder.fit_transform(x[['BUILDING CLASS CATEGORY']])

    model = RandomForestRegressor(n_estimators=100, max_depth=15, random_state=42, n_jobs=-1)
    model.fit(x, y)

    model_name = f'model_borough_{borough_code}.joblib'
    encoder_name = f'encoder_borough_{borough_code}.joblib'

    joblib.dump(model, f'models_output/{model_name}')
    joblib.dump(encoder, f'models_output/{encoder_name}')
    files.download(f'models_output/{model_name}')
    files.download(f'models_output/{encoder_name}')
  except Exception as e:
    print(f'Error {e}')
    continue
```

# **Testing**

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

MODELS_DIR = 'models_output'

print('TESTING...')
print(f"{'BOROUGH':<15} | {'R Score':<10} | {'MAE (Avg Error $)':<25} | {'RMSE (Large Error $)':<25}")
print("-" * 85)

for b_id, file_path in FILES.items():
  b_name = BOROUGH_NAMES[b_id]
  try:
    df = pd.read_excel(file_path, skiprows=4)
  except Exception as e:
    print(f'Error {e}')
    continue
  df_clean = df[
        (df['SALE PRICE'] > 10000) &
        (df['RESIDENTIAL UNITS'] > 0) &
        (df['COMMERCIAL UNITS'] == 0) &
        (df['GROSS SQUARE FEET'] > 0) &
        (df['YEAR BUILT'] > 0) &
        (df['BUILDING CLASS CATEGORY'].isin(VALID_CATEGORIES))
    ].copy()

  model_path = os.path.join(MODELS_DIR, f'model_borough_{b_id}.joblib')
  encoder_path = os.path.join(MODELS_DIR, f'encoder_borough_{b_id}.joblib')

  try:
    model = joblib.load(model_path)
    encoder = joblib.load(encoder_path)

    x = df_clean[FEATURES].copy()
    y_true = df_clean['SALE PRICE']

    x.loc[:, 'BUILDING CLASS CATEGORY'] = x['BUILDING CLASS CATEGORY'].astype(str)
    x.loc[:, 'BUILDING CLASS CATEGORY'] = encoder.transform(x[['BUILDING CLASS CATEGORY']])

    y_pred_log = model.predict(x)
    y_pred = np.expm1(y_pred_log)

    r2 = r2_score(y_true, y_pred)
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    print(f"{b_name:<15} | {r2:.2%}     | ${mae:,.0f}                | ${rmse:,.0f}")
  except Exception as e:
    print(f'Error {e}')
    continue
```
